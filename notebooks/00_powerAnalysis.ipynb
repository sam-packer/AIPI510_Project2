# 🍪 Cookie Experiment — Baseline Power Analysis (Before Pilot)

### Goal
Estimate how many cookie *pairs* (toaster vs regular oven) are required to detect a difference 
using a **paired t-test** at α = 0.05 and 80% power.

### Why this analysis?
Because there is no published quantitative study comparing toaster and regular ovens 
under home-baking conditions, we start with a **conservative assumption** of a *medium* effect size (Cohen’s d = 0.5).  
This allows us to plan a reasonable initial sample size **before** collecting any pilot data.

```python
# --- Method 1: Exact computation using statsmodels --------------------

from statsmodels.stats.power import TTestPower
from math import ceil

# Parameters
alpha = 0.05        # significance level (two-sided)
power = 0.80        # desired power (1 - β)
d = 0.5             # assumed medium effect size (Cohen's d)
tt = TTestPower()

# Compute required number of pairs (each pair = 2 cookies)
n_pairs_exact = tt.solve_power(effect_size=d, alpha=alpha, power=power, alternative='two-sided')
n_pairs_exact_ceiled = ceil(n_pairs_exact)

print(f"[Exact method]")
print(f"Required pairs (raw):     {n_pairs_exact:.2f}")
print(f"Required pairs (rounded): {n_pairs_exact_ceiled}")
print(f"Total cookies (≈):        {n_pairs_exact_ceiled * 2}")
```

### Interpretation
- The result gives the **number of pairs** (each pair = one toaster cookie + one regular cookie).  
- Always **round up** to ensure sufficient statistical power.  
- This baseline value is conservative — once we collect pilot data, 
  we will replace the assumed d=0.5 with the *observed* Cohen’s dₙ 
  to obtain a more realistic required sample size.
